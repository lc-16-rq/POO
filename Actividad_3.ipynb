```python
import tkinter as tk
from tkinter import messagebox
import statistics


class VentanaNotas:
    def __init__(self, ventana):
        self.ventana = ventana
        self.ventana.title("Notas del estudiante")
        self.entradas = []

        # Cinco casillas para las notas
        for i in range(5):
            tk.Label(ventana, text=f"Nota {i + 1}:").grid(row=i, column=0, padx=10, pady=5)
            entrada = tk.Entry(ventana)
            entrada.grid(row=i, column=1, padx=10, pady=5)
            self.entradas.append(entrada)

        # Botón
        tk.Button(ventana, text="Calcular", command=self.calcular).grid(
            row=5, column=0, columnspan=2, pady=10)

        # Parte inferior: aquí se muestran los resultados
        self.resultado = tk.Label(ventana, text="", justify="left")
        self.resultado.grid(row=6, column=0, columnspan=2, padx=10, pady=10)

    def calcular(self):
        try:
            notas = [float(e.get()) for e in self.entradas]
        except ValueError:
            messagebox.showerror("Error", "Ingresa las cinco notas como números")
            return

        promedio = statistics.mean(notas)
        desviacion = statistics.pstdev(notas)
        mayor = max(notas)
        menor = min(notas)

        self.resultado.config(text=(
            f"Promedio: {promedio:.2f}\n"
            f"Desviación estándar: {desviacion:.2f}\n"
            f"Mayor nota: {mayor}\n"
            f"Menor nota: {menor}"
        ))


def main():
    ventana = tk.Tk()
    VentanaNotas(ventana)
    ventana.mainloop()


main()
```

```python
#ejercio 2
import tkinter as tk
from tkinter import messagebox
from abc import ABC, abstractmethod
import math

class Figura(ABC):
    @abstractmethod
    def volumen(self):
        pass

    @abstractmethod
    def superficie(self):
        pass


class Cilindro(Figura):
    def __init__(self, radio, altura):
        self.radio = radio
        self.altura = altura

    def volumen(self):
        return math.pi * self.radio ** 2 * self.altura

    def superficie(self):
        return 2 * math.pi * self.radio * (self.radio + self.altura)


class Esfera(Figura):
    def __init__(self, radio):
        self.radio = radio

    def volumen(self):
        return 4 / 3 * math.pi * self.radio ** 3

    def superficie(self):
        return 4 * math.pi * self.radio ** 2


class Piramide(Figura):
    # Pirámide de base cuadrada: "base" es el lado de la base
    def __init__(self, base, altura, apotema):
        self.base = base
        self.altura = altura
        self.apotema = apotema

    def volumen(self):
        return self.base ** 2 * self.altura / 3

    def superficie(self):
        # área de la base + 4 caras triangulares (base * apotema / 2 cada una)
        return self.base ** 2 + 2 * self.base * self.apotema


class VentanaFiguras:
    def __init__(self, ventana):
        self.ventana = ventana
        self.ventana.title("Volumen y superficie de figuras")
        self.campos = {}

        self.crear_seccion("Cilindro", ["Radio (cm)", "Altura (cm)"], 0)
        self.crear_seccion("Esfera", ["Radio (cm)"], 1)
        self.crear_seccion("Pirámide", ["Base (cm)", "Altura (cm)", "Apotema (cm)"], 2)

        tk.Button(ventana, text="Calcular", command=self.calcular).grid(
            row=3, column=0, pady=10)

        # Parte inferior: resultados
        self.resultado = tk.Label(ventana, text="", justify="left")
        self.resultado.grid(row=4, column=0, padx=10, pady=10)

    def crear_seccion(self, nombre, etiquetas, fila):
        marco = tk.LabelFrame(self.ventana, text=nombre, padx=10, pady=5)
        marco.grid(row=fila, column=0, padx=10, pady=5, sticky="ew")
        self.campos[nombre] = []
        for i, texto in enumerate(etiquetas):
            tk.Label(marco, text=texto).grid(row=i, column=0, sticky="w")
            entrada = tk.Entry(marco)
            entrada.grid(row=i, column=1, padx=5, pady=2)
            self.campos[nombre].append(entrada)

    def leer(self, nombre):
        return [float(e.get()) for e in self.campos[nombre]]

    def calcular(self):
        try:
            radio_c, altura_c = self.leer("Cilindro")
            (radio_e,) = self.leer("Esfera")
            base, altura_p, apotema = self.leer("Pirámide")
        except ValueError:
            messagebox.showerror("Error", "Ingresa todos los datos como números")
            return

        if min(radio_c, altura_c, radio_e, base, altura_p, apotema) <= 0:
            messagebox.showerror("Error", "Todos los valores deben ser mayores que cero")
            return

        figuras = [
            ("Cilindro", Cilindro(radio_c, altura_c)),
            ("Esfera", Esfera(radio_e)),
            ("Pirámide", Piramide(base, altura_p, apotema)),
        ]

        texto = ""
        for nombre, figura in figuras:
            texto += (f"{nombre}\n"
                      f"  Volumen: {figura.volumen():.2f} cm³\n"
                      f"  Superficie: {figura.superficie():.2f} cm²\n\n")
        self.resultado.config(text=texto.strip())


def main():
    ventana = tk.Tk()
    VentanaFiguras(ventana)
    ventana.mainloop()


main()
```
