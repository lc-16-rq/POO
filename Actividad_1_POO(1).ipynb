```python
# Ejercio # 1

class Familia:
    def __init__(self, edad_juan):
        self.juan = edad_juan
        self.alberto = edad_juan * 2 / 3
        self.ana = edad_juan * 4 / 3
        self.mama = self.juan + self.alberto + self.ana

    def mostrar_edades(self):
        print(f"Juan:{self.juan} años")
        print(f"Alberto:{self.alberto} años")
        print(f"Ana:{self.ana} años")
        print(f"Mamá:{self.mama} años")


edad = float(input("edad de Juan: "))
familia = Familia(edad)
familia.mostrar_edades()
```

```python
# Ejercico #2

class Seguimiento:
    def __init__(self):
        self.suma = 0
        self.x = 0
        self.y = 0

    def mostrar_estado(self, linea):
        print(f"{linea:<25} SUMA={self.suma}  X={self.x}  Y={self.y}")

    def ejecutar(self):
        self.suma = 0
        self.mostrar_estado("SUMA = 0")

        self.x = 20
        self.mostrar_estado("X = 20")

        self.suma = self.suma + self.x
        self.mostrar_estado("SUMA = SUMA + X")

        self.y = 40
        self.mostrar_estado("Y = 40")

        self.x = self.x + self.y ** 2
        self.mostrar_estado("X = X + Y ** 2")

        self.suma = self.suma + self.x / self.y
        self.mostrar_estado("SUMA = SUMA + X / Y")

        print(f"\nEL VALOR DE LA SUMA ES: {self.suma}")


programa = Seguimiento()
programa.ejecutar()
```

```python
# Ejercico 3
class Empleado:
    def __init__(self, horas, valor_hora, porcentaje_retencion):
        self.horas = horas
        self.valor_hora = valor_hora
        self.porcentaje_retencion = porcentaje_retencion
        self.salario_bruto = 0
        self.retencion = 0
        self.salario_neto = 0

    def calcular(self):
        self.salario_bruto = self.horas * self.valor_hora
        self.retencion = self.salario_bruto * self.porcentaje_retencion / 100
        self.salario_neto = self.salario_bruto - self.retencion

    def mostrar_resultado(self):
        print(f"Salario bruto:  ${self.salario_bruto}")
        print(f"Retención:      ${self.retencion}")
        print(f"Salario neto:   ${self.salario_neto}")


empleado = Empleado(horas=48, valor_hora=5000, porcentaje_retencion=12.5)
empleado.calcular()
empleado.mostrar_resultado()
```

```python
# Ejercico #4
class Numero:
    def __init__(self, valor):
        self.valor = valor
        self.cuadrado = 0
        self.cubo = 0

    def calcular(self):
        self.cuadrado = self.valor ** 2
        self.cubo = self.valor ** 3

    def mostrar_resultado(self):
        print(f"Número:   {self.valor}")
        print(f"Cuadrado: {self.cuadrado}")
        print(f"Cubo:     {self.cubo}")


valor = float(input("Digite un número: "))
numero = Numero(valor)
numero.calcular()
numero.mostrar_resultado()
```

```python
# Ejercico #5
import math

class Circulo:
    def __init__(self, radio):
        self.radio = radio
        self.area = 0
        self.circunferencia = 0

    def calcular(self):
        self.area = math.pi * self.radio ** 2
        self.circunferencia = 2 * math.pi * self.radio

    def mostrar_resultado(self):
        print(f"Radio:          {self.radio}")
        print(f"Área:           {self.area:.2f}")
        print(f"Circunferencia: {self.circunferencia:.2f}")


radio = float(input("Digite el radio del círculo: "))
circulo = Circulo(radio)
circulo.calcular()
circulo.mostrar_resultado()
```
